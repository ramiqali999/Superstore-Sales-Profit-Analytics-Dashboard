```python
import pandas as pd
import numpy as np

raw = "https://raw.githubusercontent.com/yajasarora/Superstore-Sales-Analysis-with-Tableau/refs/heads/master/Superstore%20sales%20dataset.csv"
file = pd.read_csv(raw)
file.head()

file.isnull().sum()

file.drop_duplicates()

text_cols = file.select_dtypes(include='object').columns
for col in text_cols:
    file[col] = file[col].astype(str).str.strip()

file['Order Date'] = pd.to_datetime(
    file['Order Date'],
    format='%d/%m/%Y'
).dt.strftime('%d/%m/%Y')

file['Ship Date'] = pd.to_datetime(
    file['Ship Date'],
    format='%d/%m/%Y'
).dt.strftime('%d/%m/%Y')

profit = file['Profit'].fillna(0)
file.to_csv('cleaned_data.csv', index=False)

# Assuming your dataframe is loaded as 'df'
columns = file.columns

print("CREATE TABLE sales_orders (")
for col in columns:
    # Replaces spaces in column names with underscores (e.g., 'Order ID' -> 'Order_ID')
    clean_col = col.replace(" ", "_").replace("-", "_")
    print(f"    `{clean_col}` VARCHAR(255),")
print(");")


```

```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Load your clean DataFrame
df = pd.read_csv('cleaned_data.csv')

# 2. Connect to your MySQL database
# Replace 'root', 'your_password', 'localhost' (or your host IP), and 'ecommerce_db'
engine = create_engine("mysql+pymysql://root:your_password@localhost:3306/ecommerce_db")

# 3. Export all 10,000 rows directly to MySQL table
df.to_sql(name='sales_orders', con=engine, if_exists='replace', index=False)

print("Imported successfully! Total rows:", len(df))
```
